```python
import pandas as pd
import requests
from datetime import datetime, date
from icalendar import Calendar
from oauth2client.service_account import ServiceAccountCredentials
import gspread
import os

# -------- 1. Descargar eventos desde .ics --------
def fetch_ical_events(url, source_name, property_name):
    res = requests.get(url)
    cal = Calendar.from_ical(res.text)
    events = []

    for component in cal.walk():
        if component.name == "VEVENT":
            start = component.get('dtstart').dt
            end = component.get('dtend').dt
            summary = str(component.get('summary')) if component.get('summary') else ''
            events.append({
                "property_name": property_name,
                "start_date": pd.to_datetime(start),
                "end_date": pd.to_datetime(end),
                "summary": summary,
                "source": source_name
            })
    return events

# Lista de suites (asegúrate de definir esto antes si no lo has hecho)
all_events = []
for suite in suites:
    eventos = fetch_ical_events(
        url=suite['ical_url'],
        source_name=suite['source'],
        property_name=suite['nombre']
    )
    all_events.extend(eventos)

df = pd.DataFrame(all_events)
df['start_date'] = pd.to_datetime(df['start_date']).dt.date
df['end_date'] = pd.to_datetime(df['end_date']).dt.date

# -------- 2. Cargar histórico existente --------
try:
    df_hist = pd.read_csv("historico_reservas.csv", parse_dates=["start_date", "end_date"])
    df_hist["start_date"] = pd.to_datetime(df_hist["start_date"]).dt.date
    df_hist["end_date"] = pd.to_datetime(df_hist["end_date"]).dt.date
except FileNotFoundError:
    df_hist = pd.DataFrame(columns=df.columns)

# -------- 3. Actualizar histórico --------
hoy = datetime.today().date()
hist_pasadas = df_hist[(df_hist["end_date"] < hoy) | (df_hist["start_date"] < hoy)]
futuras_actualizadas = df[(df["end_date"] >= hoy) | (df["start_date"] >= hoy)]
df_final = pd.concat([hist_pasadas, futuras_actualizadas], ignore_index=True)

# -------- 4. Limpieza de nombres --------
df_final = df_final[df_final['property_name'] != 'Depto. nuevo con vista al mar']
df_final['property_name'] = df_final['property_name'].replace({
    'Bonita Suite Torres Gemelas Playa': '29-01 sur',
    'Espectacular suite Torres Gemelas Playa': '22-12 sur'
})

df_final = df_final.drop_duplicates(subset=["property_name", "start_date", "end_date", "source", "summary"])

# -------- 5. Guardar histórico actualizado --------
df_final.to_csv("historico_reservas.csv", index=False)
print("✅ Histórico actualizado correctamente.")
print(f"➡️ Total de registros: {len(df_final)}")

# -------- 6. Exportar a Google Sheets --------
scope = ["https://spreadsheets.google.com/feeds", "https://www.googleapis.com/auth/drive"]
creds = ServiceAccountCredentials.from_json_keyfile_name(
    "/Users/ricardomosqueira/airbnb_dashboard/sheets-api-calendar-0046b74b266e.json",
    scope
)
client = gspread.authorize(creds)
spreadsheet = client.open("Calendario Suites")
sheet = spreadsheet.worksheet("api python")

# Limpiar hoja
sheet.clear()

# Convertir fechas a string para evitar error
df_export = df_final.copy()
for col in df_export.columns:
    if df_export[col].dtype == 'object':
        df_export[col] = df_export[col].apply(lambda x: str(x) if isinstance(x, date) else x)
    elif 'datetime' in str(df_export[col].dtype):
        df_export[col] = df_export[col].astype(str)

# Subir a la hoja
values = [df_export.columns.tolist()] + df_export.values.tolist()
sheet.update('A1', values)
print("📤 Google Sheet actualizado exitosamente.")
```
